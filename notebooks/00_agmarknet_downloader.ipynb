```python
import pandas as pd
import time

from selenium import webdriver

from selenium.webdriver.common.by import By

from selenium.webdriver.common.keys import Keys

from selenium.webdriver.chrome.service import Service

from selenium.webdriver.support.ui import WebDriverWait

from selenium.webdriver.support import expected_conditions as EC

from webdriver_manager.chrome import ChromeDriverManager
```

```python
driver = webdriver.Chrome(
    service=Service(
        ChromeDriverManager().install()
    )
)

driver.maximize_window()

driver.get("https://agmarknet.gov.in/")
```

```python
from selenium.webdriver.common.by import By
import pandas as pd

rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")

print("Rows Found:", len(rows))

data = []

for row in rows:
    cols = row.find_elements(By.TAG_NAME, "td")
    data.append([c.text for c in cols])

df = pd.DataFrame(data)

df.head()
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time

all_data = []

while True:

    # Current page ki 10 rows read karo
    rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")

    for row in rows:
        cols = row.find_elements(By.TAG_NAME, "td")
        all_data.append([c.text for c in cols])

    print("Total Rows Collected:", len(all_data))

    # Last page check karo
    try:
        next_btn = driver.find_element(
            By.XPATH,
            "//button[@aria-label='Go to next page']"
        )

        if next_btn.get_attribute("disabled"):
            print("Last Page Reached")
            break

        # Next button click
        driver.execute_script("arguments[0].click();", next_btn)

        # Table reload hone ka wait
        time.sleep(2)

    except Exception:
        print("Finished")
        break

# DataFrame banao
df = pd.DataFrame(all_data, columns=[
    "State",
    "Commodity Group",
    "Commodity",
    "Date",
    "Arrival Quantity",
    "Arrival Unit",
    "Modal Price",
    "Price Unit"
])

print(df.head())
print(df.shape)
```
