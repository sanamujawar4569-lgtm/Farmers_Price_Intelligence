```python
import pandas as pd

files = [
    "../data/raw/potato/potato_2023.csv",
    "../data/raw/potato/potato_2024.csv",
    "../data/raw/potato/potato_2025.csv"
]

dfs = []

for file in files:

    df = pd.read_csv(file)

    # Agar first row header ban gayi ho
    if "Unnamed" in df.columns[0]:
        df.columns = df.iloc[0]
        df = df.iloc[1:].reset_index(drop=True)

    # Duplicate header rows remove
    df = df[df.iloc[:, 0] != "State/UT"]
    df = df[df.iloc[:, 3] != "Date"]

    # Standard column names
    df.columns = [
        "State",
        "Commodity_Group",
        "Commodity",
        "Date",
        "Arrival",
        "Arrival_Unit",
        "Modal_Price",
        "Price_Unit"
    ]

    dfs.append(df)

# Merge all years
df = pd.concat(dfs, ignore_index=True)

# Date
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Arrival
df["Arrival"] = (
    df["Arrival"]
      .astype(str)
      .str.replace(",", "", regex=False)
)
df["Arrival"] = pd.to_numeric(df["Arrival"], errors="coerce")

# Modal Price
df["Modal_Price"] = (
    df["Modal_Price"]
      .astype(str)
      .str.replace(",", "", regex=False)
)
df["Modal_Price"] = pd.to_numeric(df["Modal_Price"], errors="coerce")

# Footer rows remove
df = df[
    ~df["State"].astype(str).str.startswith("Note", na=False)
]

df = df[
    ~df["State"].astype(str).str.startswith("Weighted", na=False)
]

# Invalid dates remove
df = df.dropna(subset=["Date"])

# Duplicate rows remove
df = df.drop_duplicates()

# Sort
df = df.sort_values("Date").reset_index(drop=True)

print(df.shape)
print(df.head())
print(df.tail())
```

```python
df.to_csv(
    "../data/cleaned/potato_2023_2025_clean.csv",
    index=False
)

print("✅ potato_2023_2025_clean.csv saved successfully")
```

```python
import pandas as pd

files = [
    "../data/raw/wheat/wheat_2023.csv",
    "../data/raw/wheat/wheat_2024.csv",
    "../data/raw/wheat/wheat_2025.csv"
]

dfs = []

for file in files:

    df = pd.read_csv(file)

    # Agar first row header ban gayi ho
    if "Unnamed" in df.columns[0]:
        df.columns = df.iloc[0]
        df = df.iloc[1:].reset_index(drop=True)

    # Duplicate header rows remove
    df = df[df.iloc[:, 0] != "State/UT"]
    df = df[df.iloc[:, 3] != "Date"]

    # Standard column names
    df.columns = [
        "State",
        "Commodity_Group",
        "Commodity",
        "Date",
        "Arrival",
        "Arrival_Unit",
        "Modal_Price",
        "Price_Unit"
    ]

    dfs.append(df)

# Merge all years
df = pd.concat(dfs, ignore_index=True)

# Date
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Arrival
df["Arrival"] = (
    df["Arrival"]
      .astype(str)
      .str.replace(",", "", regex=False)
)
df["Arrival"] = pd.to_numeric(df["Arrival"], errors="coerce")

# Modal Price
df["Modal_Price"] = (
    df["Modal_Price"]
      .astype(str)
      .str.replace(",", "", regex=False)
)
df["Modal_Price"] = pd.to_numeric(df["Modal_Price"], errors="coerce")

# Footer rows remove
df = df[
    ~df["State"].astype(str).str.startswith("Note", na=False)
]

df = df[
    ~df["State"].astype(str).str.startswith("Weighted", na=False)
]

# Invalid dates remove
df = df.dropna(subset=["Date"])

# Duplicate rows remove
df = df.drop_duplicates()

# Sort
df = df.sort_values("Date").reset_index(drop=True)

print(df.shape)
print(df.head())
print(df.tail())
```

```python
df.to_csv(
    "../data/cleaned/wheat_2023_2025_clean.csv",
    index=False
)

print("✅ rice_2023_2025_clean.csv saved successfully")
```
