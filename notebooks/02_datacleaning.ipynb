```python
import pandas as pd
import os
```

```python
PROJECT_PATH = r"C:\Users\DELL\Desktop\Farmers_Price_Intelligence"

PROCESSED_PATH = os.path.join(PROJECT_PATH, "data", "processed")

CLEANED_PATH = os.path.join(PROJECT_PATH, "data", "cleaned")

os.makedirs(CLEANED_PATH, exist_ok=True)
```

```python
datasets = {}

crop_names = [
    "onion",
    "tomato",
    "potato",
    "wheat",
    "rice"
]

for crop in crop_names:

    datasets[crop] = pd.read_csv(
        os.path.join(PROCESSED_PATH, f"{crop}.csv")
    )

print("Datasets Loaded")
```

```python
def clean_dataset(df):

    # Rename date column
    if "t" in df.columns:
        df.rename(columns={"t": "date"}, inplace=True)

    # Convert date
    df["date"] = pd.to_datetime(df["date"])

    # Remove duplicates
    df.drop_duplicates(inplace=True)

    # Remove missing values
    df.dropna(inplace=True)

    # Remove invalid prices
    df = df[
        (df["p_min"] <= df["p_modal"]) &
        (df["p_modal"] <= df["p_max"])
    ]

    df.reset_index(drop=True, inplace=True)

    return df
```

```python
for crop in datasets:

    datasets[crop] = clean_dataset(
        datasets[crop]
    )

print("Cleaning Completed")
```
